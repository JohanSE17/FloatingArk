```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Simulated wave condition data (height in meters, label: 0=moderada, 1=extrema)
X = np.array([[1.0], [1.5], [2.0], [2.5], [3.0], [3.5], [4.0]])
y = np.array([0, 0, 0, 1, 1, 1, 1])

# Train Naive Bayes model
model = GaussianNB()
model.fit(X, y)
y_pred = model.predict(X)

# Confusion Matrix
cm = confusion_matrix(y, y_pred)
plt.figure(figsize=(8, 6), dpi=300)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Moderada', 'Extrema'], yticklabels=['Moderada', 'Extrema'])
plt.title('Matriz de Confusión - Clasificación de Condiciones de Oleaje')
plt.ylabel('Verdadero')
plt.xlabel('Predicho')
plt.savefig('img/naive_bayes_confusion.png', bbox_inches='tight')
plt.close()
```
